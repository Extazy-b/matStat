```python
import pandas as pd
import numpy as np
from scipy.stats import t
```

```python
MY_VARIANT = 5
df = pd.read_csv('../data/2.csv', header=None, names=['variant', 'country', 'value'])
```

## 1. Выбрать данные своего варианта

```python
data = df[df['variant'] == MY_VARIANT].copy()
n = len(data)
print(f"Количество наблюдений в варианте {MY_VARIANT}: {n}")
countries = data['country'].unique()
print(f"Страны в варианте {MY_VARIANT}: {countries}")
data.head(5)
```

### Функция интервалов

```python
def confidence_intervals(sample, alpha, mode="center"):
    n = len(sample)
    mean = np.mean(sample)
    std_err = np.std(sample, ddof=1) / np.sqrt(n)
    if mode == "center":
        quant = t.ppf(1 - alpha/2, df=n-1)
        return mean - quant * std_err, mean + quant * std_err
    if mode == "left":
        return mean + t.ppf(1 - alpha, df=n-1) * std_err
    if mode == "right":
        return mean - t.ppf(1 - alpha, df=n-1) * std_err
    return
```

```python
results = {}
for country in countries:
    sample = data[data['country'] == country]['value'].values
    results[country] = {
        'sample_size': len(sample),
        'mean': np.mean(sample),
        'std': np.std(sample, ddof=1),
        '90%_sym_CI': confidence_intervals(sample, 0.1),
        '95%_left_CI_upper': confidence_intervals(sample, 0.05, "left")
    }
```

```python
results
```

```python
for country, res in results.items():
    print(f"Страна: {country}")
    print(f"Размер выборки: {res['sample_size']}")
    print(f"Выборочное среднее: {res['mean']:.4f}")
    print(f"Выборочное ст.откл.: {res['std']:.4f}")
    print(f"90%-й симметричный ДИ для среднего: ({res['90%_sym_CI'][0]:.4f}, {res['90%_sym_CI'][1]:.4f})")
    print(f"95%-й левосторонний ДИ (верхняя граница): (-∞, {res['95%_left_CI_upper']:.4f}]")
```
